```python
# ============================================================
# DATA CLEANING & VISUALIZATION PROJECT
# Real-Time Weather Data Analysis using Open-Meteo API
# ============================================================

# If libraries are missing, uncomment and run this:
# !pip install pandas numpy requests matplotlib seaborn

import pandas as pd
import numpy as np
import requests
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
import os

sns.set_theme(style="whitegrid")

print("=" * 70)
print("       DATA CLEANING & VISUALIZATION PROJECT")
print("       REAL-TIME WEATHER DATA ANALYSIS")
print("=" * 70)


# ============================================================
# 1. FETCH REAL-WORLD DATA
# ============================================================

print("\n[1] Fetching weather data...")

# Chennai coordinates
LATITUDE = 13.0827
LONGITUDE = 80.2707

API_URL = "https://api.open-meteo.com/v1/forecast"

PARAMS = {
    "latitude": LATITUDE,
    "longitude": LONGITUDE,
    "hourly": (
        "temperature_2m,"
        "relative_humidity_2m,"
        "precipitation,"
        "wind_speed_10m"
    ),
    "forecast_days": 7,
    "timezone": "Asia/Kolkata"
}

try:
    response = requests.get(
        API_URL,
        params=PARAMS,
        timeout=30
    )

    response.raise_for_status()

    weather_data = response.json()

    print("✓ Data fetched successfully!")

except Exception as e:
    print("❌ Error fetching data:", e)
    raise


# ============================================================
# 2. CREATE DATAFRAME
# ============================================================

hourly = weather_data["hourly"]

df = pd.DataFrame(hourly)

df.rename(
    columns={
        "time": "datetime",
        "temperature_2m": "temperature",
        "relative_humidity_2m": "humidity",
        "wind_speed_10m": "wind_speed"
    },
    inplace=True
)

df["datetime"] = pd.to_datetime(df["datetime"])

print("\n[2] Dataset created.")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

display(df.head())


# ============================================================
# 3. DATA INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("DATA INFORMATION")
print("=" * 70)

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nStatistical summary:")
display(df.describe())


# ============================================================
# 4. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

missing = df.isnull().sum()

print("\nMissing values:")
display(missing)

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

print("\nMissing percentage:")
display(missing_percentage)


# ============================================================
# 5. HANDLE MISSING VALUES
# ============================================================

numeric_columns = [
    "temperature",
    "humidity",
    "precipitation",
    "wind_speed"
]

for column in numeric_columns:

    if df[column].isnull().sum() > 0:

        df[column] = df[column].fillna(
            df[column].median()
        )

print("✓ Missing values handled.")

print("\nMissing values after cleaning:")
display(df.isnull().sum())


# ============================================================
# 6. DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("DUPLICATE ANALYSIS")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)

df.drop_duplicates(inplace=True)

print(
    "Duplicate rows after cleaning:",
    df.duplicated().sum()
)


# ============================================================
# 7. FEATURE ENGINEERING
# ============================================================

print("\n" + "=" * 70)
print("FEATURE ENGINEERING")
print("=" * 70)

df["date"] = df["datetime"].dt.date

df["hour"] = df["datetime"].dt.hour

df["day"] = df["datetime"].dt.day_name()

df["day_number"] = df["datetime"].dt.day

display(df.head())


# ============================================================
# 8. OUTLIER DETECTION USING IQR
# ============================================================

print("\n" + "=" * 70)
print("OUTLIER DETECTION")
print("=" * 70)


def detect_outliers(data, column):

    Q1 = data[column].quantile(0.25)

    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return outliers, lower_bound, upper_bound


outlier_summary = {}

for column in numeric_columns:

    outliers, lower, upper = detect_outliers(
        df,
        column
    )

    outlier_summary[column] = len(outliers)

    print(
        f"{column:15} → "
        f"{len(outliers)} outliers"
    )


# ============================================================
# 9. OUTLIER VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df[numeric_columns]
)

plt.title(
    "Weather Variables - Outlier Detection"
)

plt.xlabel("Variables")

plt.ylabel("Values")

plt.xticks(rotation=20)

plt.tight_layout()

plt.show()


# ============================================================
# 10. TEMPERATURE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("TEMPERATURE ANALYSIS")
print("=" * 70)

average_temperature = df["temperature"].mean()

maximum_temperature = df["temperature"].max()

minimum_temperature = df["temperature"].min()

print(
    f"Average temperature : "
    f"{average_temperature:.2f} °C"
)

print(
    f"Maximum temperature : "
    f"{maximum_temperature:.2f} °C"
)

print(
    f"Minimum temperature : "
    f"{minimum_temperature:.2f} °C"
)


# ============================================================
# 11. TEMPERATURE TREND
# ============================================================

plt.figure(figsize=(15, 6))

plt.plot(
    df["datetime"],
    df["temperature"],
    marker="o",
    linewidth=2
)

plt.title(
    "7-Day Temperature Forecast"
)

plt.xlabel("Date and Time")

plt.ylabel("Temperature (°C)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 12. HUMIDITY ANALYSIS
# ============================================================

average_humidity = df["humidity"].mean()

print(
    f"\nAverage humidity: "
    f"{average_humidity:.2f}%"
)

plt.figure(figsize=(15, 6))

plt.plot(
    df["datetime"],
    df["humidity"],
    marker="o",
    linewidth=2
)

plt.title(
    "7-Day Humidity Forecast"
)

plt.xlabel("Date and Time")

plt.ylabel("Humidity (%)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 13. PRECIPITATION ANALYSIS
# ============================================================

total_rainfall = df["precipitation"].sum()

maximum_rainfall = df["precipitation"].max()

print(
    f"\nTotal precipitation: "
    f"{total_rainfall:.2f} mm"
)

print(
    f"Maximum hourly precipitation: "
    f"{maximum_rainfall:.2f} mm"
)

plt.figure(figsize=(15, 6))

plt.bar(
    df["datetime"],
    df["precipitation"]
)

plt.title(
    "Hourly Precipitation Forecast"
)

plt.xlabel("Date and Time")

plt.ylabel("Precipitation (mm)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 14. WIND SPEED ANALYSIS
# ============================================================

average_wind = df["wind_speed"].mean()

maximum_wind = df["wind_speed"].max()

print(
    f"\nAverage wind speed: "
    f"{average_wind:.2f} km/h"
)

print(
    f"Maximum wind speed: "
    f"{maximum_wind:.2f} km/h"
)

plt.figure(figsize=(15, 6))

plt.plot(
    df["datetime"],
    df["wind_speed"],
    marker="o",
    linewidth=2
)

plt.title(
    "7-Day Wind Speed Forecast"
)

plt.xlabel("Date and Time")

plt.ylabel("Wind Speed (km/h)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 15. DAILY TEMPERATURE ANALYSIS
# ============================================================

daily_temperature = (
    df.groupby("date")
    .agg(
        average_temperature=("temperature", "mean"),
        maximum_temperature=("temperature", "max"),
        minimum_temperature=("temperature", "min")
    )
    .reset_index()
)

print("\nDaily temperature:")
display(daily_temperature)


plt.figure(figsize=(12, 6))

plt.plot(
    daily_temperature["date"],
    daily_temperature["average_temperature"],
    marker="o",
    label="Average"
)

plt.plot(
    daily_temperature["date"],
    daily_temperature["maximum_temperature"],
    marker="o",
    label="Maximum"
)

plt.plot(
    daily_temperature["date"],
    daily_temperature["minimum_temperature"],
    marker="o",
    label="Minimum"
)

plt.title(
    "Daily Temperature Analysis"
)

plt.xlabel("Date")

plt.ylabel("Temperature (°C)")

plt.xticks(rotation=45)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 16. DAILY RAINFALL
# ============================================================

daily_rainfall = (
    df.groupby("date")["precipitation"]
    .sum()
    .reset_index()
)

print("\nDaily rainfall:")
display(daily_rainfall)


plt.figure(figsize=(12, 6))

plt.bar(
    daily_rainfall["date"],
    daily_rainfall["precipitation"]
)

plt.title(
    "Daily Rainfall"
)

plt.xlabel("Date")

plt.ylabel("Rainfall (mm)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 17. HOURLY TEMPERATURE PATTERN
# ============================================================

hourly_average = (
    df.groupby("hour")
    .agg(
        average_temperature=("temperature", "mean"),
        average_humidity=("humidity", "mean"),
        average_wind_speed=("wind_speed", "mean")
    )
    .reset_index()
)

print("\nHourly weather pattern:")
display(hourly_average)


plt.figure(figsize=(12, 6))

plt.plot(
    hourly_average["hour"],
    hourly_average["average_temperature"],
    marker="o"
)

plt.title(
    "Average Temperature by Hour"
)

plt.xlabel("Hour of Day")

plt.ylabel("Temperature (°C)")

plt.xticks(range(24))

plt.tight_layout()

plt.show()


# ============================================================
# 18. TEMPERATURE DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    df["temperature"],
    kde=True
)

plt.title(
    "Temperature Distribution"
)

plt.xlabel("Temperature (°C)")

plt.ylabel("Frequency")

plt.tight_layout()

plt.show()


# ============================================================
# 19. TEMPERATURE VS HUMIDITY
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="humidity",
    y="temperature"
)

plt.title(
    "Temperature vs Humidity"
)

plt.xlabel("Humidity (%)")

plt.ylabel("Temperature (°C)")

plt.tight_layout()

plt.show()


# ============================================================
# 20. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("CORRELATION ANALYSIS")
print("=" * 70)

correlation = df[
    numeric_columns
].corr()

display(correlation)


plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title(
    "Weather Variables Correlation"
)

plt.tight_layout()

plt.show()


# ============================================================
# 21. HOTTEST AND COLDEST OBSERVATIONS
# ============================================================

hottest = df.loc[
    df["temperature"].idxmax()
]

coldest = df.loc[
    df["temperature"].idxmin()
]

print("\n" + "=" * 70)
print("EXTREME WEATHER OBSERVATIONS")
print("=" * 70)

print("\n🔥 Hottest observation:")

display(
    hottest[
        [
            "datetime",
            "temperature",
            "humidity",
            "precipitation",
            "wind_speed"
        ]
    ].to_frame().T
)

print("\n❄️ Coldest observation:")

display(
    coldest[
        [
            "datetime",
            "temperature",
            "humidity",
            "precipitation",
            "wind_speed"
        ]
    ].to_frame().T
)


# ============================================================
# 22. FINAL DATASET
# ============================================================

print("\n" + "=" * 70)
print("FINAL CLEANED DATASET")
print("=" * 70)

display(df.head(10))

print(
    f"\nFinal dataset size: "
    f"{df.shape[0]} rows × {df.shape[1]} columns"
)


# ============================================================
# 23. SAVE CLEANED DATASET
# ============================================================

output_file = "cleaned_weather_data.csv"

df.to_csv(
    output_file,
    index=False
)

print(
    f"\n✓ Cleaned dataset saved as: "
    f"{output_file}"
)


# ============================================================
# 24. FINAL PROJECT REPORT
# ============================================================

print("\n")
print("=" * 70)
print("                 FINAL PROJECT REPORT")
print("=" * 70)

print(
    f"""
Dataset:
    Real-world weather forecast data

Location:
    Chennai, India

Observations:
    {len(df)}

Average Temperature:
    {df["temperature"].mean():.2f} °C

Maximum Temperature:
    {df["temperature"].max():.2f} °C

Minimum Temperature:
    {df["temperature"].min():.2f} °C

Average Humidity:
    {df["humidity"].mean():.2f} %

Total Precipitation:
    {df["precipitation"].sum():.2f} mm

Average Wind Speed:
    {df["wind_speed"].mean():.2f} km/h

Duplicate Rows Removed:
    {duplicate_count}

Outliers:
    {sum(outlier_summary.values())} observations detected

Missing Values:
    Handled using median imputation

Cleaning:
    ✓ Missing values
    ✓ Duplicate records
    ✓ Datetime conversion
    ✓ Feature engineering
    ✓ IQR outlier detection

Analysis:
    ✓ Descriptive statistics
    ✓ Temperature analysis
    ✓ Humidity analysis
    ✓ Rainfall analysis
    ✓ Wind analysis
    ✓ Correlation analysis
    ✓ Daily analysis
    ✓ Hourly analysis

Visualization:
    ✓ Line charts
    ✓ Bar charts
    ✓ Histogram
    ✓ Scatter plot
    ✓ Box plot
    ✓ Correlation heatmap

Output:
    {output_file}
"""
)

print("=" * 70)
print("             PROJECT COMPLETED SUCCESSFULLY ✓")
print("=" * 70)
```
